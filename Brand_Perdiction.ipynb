## Generate Synthetic Dataset for Car Brand Prediction

### Subtask:
Create a synthetic dataset with at least 1000 samples. Include 'Mileage', 'Age', 'Engine Size', and 'Price' as numerical features, and 'Brand' (e.g., 'Toyota', 'Honda', 'BMW', 'Mercedes') as the categorical target variable. Ensure plausible relationships between these features and the car brand, with some random noise.


**Reasoning**:
To begin generating the synthetic dataset, I need to import the necessary libraries, define the car brands, and set the number of samples as instructed. This sets up the foundational elements for the data generation process.



```python
import numpy as np
import pandas as pd

# Define a list of car brands
brands = ['Toyota', 'Honda', 'BMW', 'Mercedes']

# Set the number of samples for the dataset
n_samples = 1000

print(f"Libraries imported: numpy, pandas")
print(f"Car brands defined: {brands}")
print(f"Number of samples set: {n_samples}")
```

**Reasoning**:
Now that the foundational elements (`brands`, `n_samples`) are defined, I will proceed to generate the synthetic car data. This involves iterating `n_samples` times, randomly selecting a brand, and then generating feature values (Mileage, Age, Engine Size, Price) based on brand-specific plausible distributions with added noise, as specified in the instructions. Finally, I will convert the collected data into a Pandas DataFrame.



```python
car_data = []

# Define parameters for each brand to ensure plausible relationships and noise
brand_params = {
    'Toyota': {'Mileage': (70000, 15000), 'Age': (5, 2), 'Engine Size': (1.8, 0.3), 'Price': (20000, 5000)},
    'Honda': {'Mileage': (65000, 12000), 'Age': (4, 1.5), 'Engine Size': (1.7, 0.2), 'Price': (22000, 4500)},
    'BMW': {'Mileage': (50000, 10000), 'Age': (3, 1), 'Engine Size': (2.5, 0.5), 'Price': (45000, 10000)},
    'Mercedes': {'Mileage': (48000, 9000), 'Age': (3, 1), 'Engine Size': (2.8, 0.6), 'Price': (50000, 12000)}
}

for _ in range(n_samples):
    brand = np.random.choice(brands)
    params = brand_params[brand]

    # Generate features with randomness, ensuring plausible ranges
    mileage = max(1000, np.random.normal(params['Mileage'][0], params['Mileage'][1]))
    age = max(0.5, np.random.normal(params['Age'][0], params['Age'][1]))
    engine_size = max(1.0, np.random.normal(params['Engine Size'][0], params['Engine Size'][1]))
    price = max(5000, np.random.normal(params['Price'][0], params['Price'][1]))

    car_data.append({
        'Brand': brand,
        'Mileage': round(mileage, 0),
        'Age': round(age, 1),
        'Engine Size': round(engine_size, 2),
        'Price': round(price, 2)
    })

# Convert the list of dictionaries into a Pandas DataFrame
df_cars = pd.DataFrame(car_data)

print(f"Synthetic dataset created with {len(df_cars)} samples.")
print("First 5 rows of the dataset:")
print(df_cars.head())
print("\nDataset info:")
df_cars.info()
```
